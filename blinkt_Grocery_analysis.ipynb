```python
#Compare better-performing and lower-performing products and outlets

item_sales = df.groupby('Item Type')['Sales'].agg(Total_Sales='sum',Average_Sales='mean',Number_of_Records='count')

print(item_sales.sort_values('Average_Sales', ascending=False))
```

```python
product_sales = df.groupby('Item Identifier')['Sales'].agg(
    Total_Sales='sum',
    Average_Sales='mean',
    Number_of_Records='count'
)

print("Top 10 products:")
print(product_sales.sort_values('Total_Sales', ascending=False).head(10))

print("\nBottom 10 products:")
print(product_sales.sort_values('Total_Sales', ascending=True).head(10))
```

```python
outlet_performance = df.groupby('Outlet Identifier')['Sales'].agg(
    Total_Sales='sum',
    Average_Sales='mean',
    Number_of_Records='count'
)

print("Top 10 outlets:")
print(outlet_performance.sort_values('Total_Sales', ascending=False).head(10))

print("\nBottom 10 outlets:")
print(outlet_performance.sort_values('Total_Sales', ascending=True).head(10))
```

```python
outlet_performance['Total_Sales'].sort_values().plot(
    kind='bar', figsize=(10, 5)
)

plt.title('Total Sales by Outlet')
plt.xlabel('Outlet Identifier')
plt.ylabel('Total Sales')
plt.tight_layout()
plt.show()
```

```python
#Identify products contributing more to sales 
product_sales = df.groupby('Item Type')['Sales'].sum().sort_values(ascending=False)

print(product_sales)
```

```python
#Top 10 product categories
product_sales.head(10).plot(kind='bar', figsize=(10, 5))
plt.title('Top Product Categories by Total Sales')
plt.xlabel('Item Type')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#Calculate each category's contribution to total sales
contribution = (product_sales / df['Sales'].sum() * 100).round(2)

print(contribution.head(10))
```

```python
#Calculate each category's contribution to total sales
contribution = (product_sales / df['Sales'].sum() * 100).round(2)

print(contribution.head(10))
```

```python
#Identify better-performing outlets and locations
```

```python
outlet_type_sales = df.groupby('Outlet Type')['Sales'].agg(  #Compare outlet types
    Total_Sales='sum',
    Average_Sales='mean',
    Transactions='count'
).sort_values('Total_Sales', ascending=False)

print(outlet_type_sales)
```

```python
location_sales = df.groupby('Outlet Location Type')['Sales'].agg(  #Compare outlet locations
    Total_Sales='sum',
    Average_Sales='mean',
    Transactions='count'
).sort_values('Total_Sales', ascending=False)

print(location_sales)
```

```python
outlet_sales = df.groupby('Outlet Identifier')['Sales'].agg(      #Compare individual outlets
    Total_Sales='sum',
    Average_Sales='mean',
    Transactions='count'
).sort_values('Total_Sales', ascending=False)

print(outlet_sales)
```

```python
#Find low-performing product categories
category_performance = df.groupby('Item Type')['Sales'].agg(
    Total_Sales='sum',
    Average_Sales='mean',
    Count='count'
)

print(category_performance.sort_values('Average_Sales').head(10))
```

```python
#Find low-performing outlets
print(outlet_sales.sort_values('Total_Sales').head(10))
```

```python
#Check product visibility and sales 
df['Visibility Group'] = pd.cut(
    df['Item Visibility'],
    bins=[-0.01, 0.05, 0.10, 0.20, 1],
    labels=['Low', 'Medium', 'High', 'Very High']
)

print(df.groupby('Visibility Group', observed=True)['Sales'].mean())
```

```python
#Check outlet size and sales
print(
    df.groupby('Outlet Size', dropna=False)['Sales']
      .agg(['sum', 'mean', 'count'])
      .sort_values('mean', ascending=False)
)
```

```python
total_sales = df['Sales'].sum()
average_sales = df['Sales'].mean()
total_records = len(df)

fig, ax = plt.subplots(1, 3, figsize=(15, 4))

# Total sales
ax[0].bar(['Total Sales'], [total_sales])
ax[0].set_title('Overall Total Sales')
ax[0].set_ylabel('Sales')

# Average sales
ax[1].bar(['Average Sales'], [average_sales])
ax[1].set_title('Average Sales per Record')

# Number of records
ax[2].bar(['Records'], [total_records])
ax[2].set_title('Total Records')

plt.suptitle('BlinkIT Grocery Business — Overall Performance')
plt.tight_layout()
plt.show()
```

```python
suggestions to improve business performance
1. Focus on top-selling products
Maintain sufficient stock for high-revenue categories, monitor demand, and avoid stockouts.
2. Improve lower-performing outlets
Compare inventory availability, pricing, product mix, and operating costs with successful outlets before making changes.
3. Adapt to location demand
Adjust product assortment and promotions to match the preferences of customers in each location tier.
4. Test product visibility improvements
Experiment with shelf placement and product presentation, then measure whether sales improve.
5. Monitor key performance indicators
Track total sales, average sales per record, category contribution, stockouts, and outlet-level trends regularly.
```
